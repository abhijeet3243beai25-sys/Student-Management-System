```python
d2=np.array([[11,22,25,26,29],[100,120,150,160,130],[74,87,89,96,66]])
d2[d2>50]
```

```python
d2=np.where(d2==150,1500,d2)
d2
```

```python
q1=float(np.percentile(d2,25))
q3=float(np.percentile(d2,75))
iqr=q3-q1
lf=q1-(1.5*iqr)
uf=q3+(1.5*iqr)
print(lf)
print(uf)
```

```python
n=int(input())
arr=[]
for i in range(n):
    num=int(input())
    arr.append(num)
Mean=np.mean(arr)
Median=np.median(arr)
q1=float(np.percentile(arr,25))
q3=float(np.percentile(arr,75))
iqr=q3-q1
print("Mean: ",Mean)
print("Median: ",Median)
print("q1: ",q1)
print("q3: ",q3)
print("iqr: ",iqr)
```

```python
no=int(input())
markslist=[]
for i in range(no):
    marks=float(input())
    markslist.append(marks)
q1=np.percentile(markslist,25)
q3=np.percentile(markslist,75)
iqr=q3-q1
lf=q1-(1.5*iqr)
uf=q3+(1.5*iqr)
for i in markslist:
    if i<lf or i>uf:
        continue
    else:
        print(i)
```
