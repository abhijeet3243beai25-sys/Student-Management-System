how to calculate new column

```python
df["package"]=df["Salary"]*12
```

```python
df["per_day"]=df["package"]/365
```

```python
def n(x):
    if x=="A":
        return "Amit"
    elif x=="B":
        return "Bhanu"
    elif x=="C":
        return "Charan"
    else:
        return "Dev"
```

```python
df["name"]=df["name"].apply(n)
df
```

```python

def n(x):
    if x<=50000:
        return "low"
    else:
        return "high"
```

```python
df["view_salary"]=df["Salary"].apply(n)
df
```

```python
df.drop(columns="view_salary",inplace=True)
```

```python
df
```

```python
df.drop(columns=["per_day"],inplace=True)
```
